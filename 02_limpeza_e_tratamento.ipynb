# Polimento 

```python
df[["price"]] = df[["price"]].replace("[^0-9.]","", regex=True,).astype(float)
```

###### colunas com inconsistências 

```python
print(df["shopping_mall"].unique())
```

```python
df["shopping_mall"] = df["shopping_mall"].str.lower()

ajustar = {
'kanyon': 'Kanyon',
'forum istanbul': 'Forum Istanbul' ,
'metrocity avm': 'Metrocity AVM',
'metropol avm' : 'Metropol İstanbul', 
'istinye park' :  'Istinye Park',
'mall of istanbul': 'Mall of Istanbul',
'emaar square mall' : 'Emaar Square Mall',
'cevahir avm' : 'Cevahir Shopping',
'viaport outlet' : 'Viaport Outlet',
'zorlu center' : 'Zorlu Center' 
}

df["shopping_mall"] = df["shopping_mall"].replace(ajustar)
```

```python
print(df["shopping_mall"].unique())
```

```python
df["shopping_mall"] = df["shopping_mall"].str.capitalize()
```

```python
print(df['shopping_mall'].unique())
```

###### ajustando formato de datas 

```python
df["invoice_date"] = pd.to_datetime(df["invoice_date"], format= "mixed", dayfirst = True)
```
